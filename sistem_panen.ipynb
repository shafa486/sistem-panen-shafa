```python
# Fungsi Hitung Hasil Panen
def hitung_panen(hasil_kg):
    print(f"Total Hasil Panen: {hasil_kg} kg")

# Fungsi Perhitungan Diskon
def hitung_diskon(harga_total, persentase):
    return harga_total - (harga_total * (persentase / 100))

# Uji Coba Program
hitung_panen(500)
total_setelah_diskon = hitung_diskon(1000000, 10)
print(f"Harga setelah diskon 10%: Rp{total_setelah_diskon:,.0f}")
```
